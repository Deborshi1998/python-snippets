# Group by and other operations


```python
df_salary = spark.read.format("csv") \
          .option("header", "true") \
          .option("inferSchema", "true") \
          .load("/content/sample_data/Salary_Data.csv")
```

```python
df_salary.describe().show()
```

```python
df_salary.show(10)
```

```python
#Groupby
df_salary.groupBy("Gender").agg(fun.avg("Salary")).show()
```

```python
(df_salary.groupBy("Gender").avg('Salary')).alias('sum_val').show()
```

```python


df_chained = (
    df_salary.filter(col("Education Level") == "Master's")
    .groupBy(col("Gender"))
    .agg(fun.avg(col("Salary")).alias("avg_sal_val"))
    .sort(col("Gender").desc())
)

display(df_chained)
```

```python
df_chained.show()
```
